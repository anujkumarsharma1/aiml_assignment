```python
# Simple Reflex Agent
courses = {
    "webdev": ("Web Development", "12 weeks", "Rs.8999", "HTML, CSS, JS, React"),
    "genai": ("Generative AI", "8 weeks", "Rs.12999", "Prompt Engineering, RAG, LLMs"),
    "ml": ("Machine Learning", "10 weeks", "Rs.10999", "Python, Regression, Neural Networks"),
}
HELP = "Ask me about Web Dev, Gen-AI, or ML courses."
ASK = "Please tell me which course - Web Dev, Gen-AI, or ML."
SORRY = "Sorry, I didn't understand. Try asking about Web Dev, Gen-AI, or ML."
QUERY_WORDS = ["price", "cost", "duration", "how long", "curriculum", "learn"]

def find_course(text):
    if "web dev" in text or "webdev" in text:
        return "webdev"
    if any(k in text for k in ["gen ai", "genai", "gen-ai", "generative ai"]):
        return "genai"
    if "machine learning" in text or text == "ml":
        return "ml"

def reflex_response(text):
    # Actuator
    if "hi" in text or "hello" in text:
        return "Hi! " + HELP
    if "courses" in text:
        return "We offer 3 courses - Web Dev, Gen-AI, and Machine Learning."
    course = find_course(text)
    if course:
        name, weeks, price, topics = courses[course]
        return f"{name} - {weeks}, {price}. Topics: {topics}."
    if any(k in text for k in QUERY_WORDS):
        return ASK
    return SORRY
```

```python
# Model Based Agent
memory = {"current_course": None, "user_name": None}

def model_response(text, memory):
    # Actuator
    if "my name is" in text:
        name = text.split("my name is")[-1].strip()
        memory["user_name"] = name
        return f"Nice to meet you, {name}!"
    if "hi" in text or "hello" in text:
        user = memory["user_name"]
        return f"Hi {user}! " + HELP if user else "Hi! " + HELP
    if "courses" in text:
        return "We offer 3 courses - Web Dev, Gen-AI, and Machine Learning."
    course = find_course(text)
    if course:
        memory["current_course"] = course
        name, weeks, price, topics = courses[course]
        return f"{name} - {weeks}, {price}. Topics: {topics}."
    course = memory["current_course"]
    if course is None:
        return ASK if any(k in text for k in QUERY_WORDS) else SORRY
    name, weeks, price, topics = courses[course]
    if "price" in text or "cost" in text:
        return f"{name} costs {price}."
    if "duration" in text or "how long" in text:
        return f"{name} takes {weeks}."
    if "curriculum" in text or "learn" in text:
        return f"{name} covers {topics}."
    return SORRY
```

```python
# Environment
mode = "reflex"
print("Bot: Welcome! I'm your course assistant.")
print("Bot: Type 'toggle' to switch between Reflex and Model-Based mode. Type 'exit' to leave.")

while True:
    # Sensor
    text = input("You : ").lower()
    if text in ("exit", "quit", "bye"):
        print("Bot: Goodbye!")
        break
    elif text == "toggle":
        mode = "model" if mode == "reflex" else "reflex"
        print(f"Bot: Switched to {mode} mode.")
    elif mode == "reflex":
        print("Bot:", reflex_response(text))
    else:
        print("Bot:", model_response(text, memory))
```
