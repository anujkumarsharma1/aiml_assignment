### Forward Chaining

```python
def evaluate_rule_fc(condition, known):
    temp = condition.replace("(", " ( ").replace(")", " ) ")
    words = temp.split()

    new_words = []
    for w in words:
        if w in ("and", "or", "(", ")"):
            new_words.append(w)
        elif w in known:
            new_words.append("True")
        else:
            new_words.append("False")

    expression = " ".join(new_words)
    return eval(expression)


def forward_chaining(rules, facts, goals):
    known = list(facts)
    changed = True

    while changed:
        changed = False

        for condition, conclusion in rules:
            rule_is_true = evaluate_rule_fc(condition, known)

            if rule_is_true and conclusion not in known:
                known.append(conclusion)
                print("Rule fired:", condition, "->", conclusion)
                changed = True

    # Automatically derive and return all matching goals from the case study
    derived_goals = [g for g in goals if g in known]
    return derived_goals
```

### Backward Chaining

```python
def evaluate_rule_bc(condition, rules, facts):
    temp = condition.replace("(", " ( ").replace(")", " ) ")
    words = temp.split()

    new_words = []
    for w in words:
        if w in ("and", "or", "(", ")"):
            new_words.append(w)
        elif backward_chaining(rules, facts, w):
            new_words.append("True")
        else:
            new_words.append("False")

    expression = " ".join(new_words)
    return eval(expression)


def backward_chaining(rules, facts, goal):
    if goal in facts:
        return True

    for condition, conclusion in rules:
        if conclusion == goal:
            if evaluate_rule_bc(condition, rules, facts):
                return True

    return False
```

```python
def input_facts(facts):
    print("\nAVAILABLE FACTS")
    for i, fact in enumerate(facts, 1):
        print(i, fact)

    choice = input("Enter fact numbers separated by commas: ")
    return [facts[int(x.strip()) - 1] for x in choice.split(",")]


def input_goal(goals):
    print("\nAVAILABLE GOALS")
    for i, goal in enumerate(goals, 1):
        print(i, goal)

    choice = int(input("Enter goal number: "))
    return goals[choice - 1]
```
