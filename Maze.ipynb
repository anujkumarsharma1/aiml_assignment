```python
import time
from collections import deque
```

```python
def get_maze():
    print("\n--- MAZE INPUT ---")
    print("Use 0 = open path, 1 = wall, S = start, E = end")
    rows = int(input("Enter number of rows: "))
    cols = int(input("Enter number of columns: "))

    maze = []
    for i in range(rows):
        row = input(f"Row {i+1}: ").replace(" ", "")
        while len(row) != cols:
            print(f"Row must have exactly {cols} characters, try again.")
            row = input(f"Row {i+1}: ").replace(" ", "")
        maze.append(row)

    start = end = None
    for r in range(rows):
        for c in range(cols):
            if maze[r][c] == "S":
                start = (r, c)
            elif maze[r][c] == "E":
                end = (r, c)
    return maze, start, end


def get_neighbours(maze, pos):
    r, c = pos
    moves = [(r - 1, c), (r + 1, c), (r, c - 1), (r, c + 1)]
    return [(nr, nc) for nr, nc in moves
            if 0 <= nr < len(maze) and 0 <= nc < len(maze[nr]) and maze[nr][nc] != "1"]


def show_maze_path(maze, path):
    if not path:
        print("No path found!")
        return
    grid = [list(row) for row in maze]
    for r, c in path:
        if grid[r][c] == "0":
            grid[r][c] = "*"
    for row in grid:
        print("".join(row))
```

```python
def search(maze, start, end, bfs):
    todo = deque([[start]])      # stores paths
    visited = set()
    explored = 0
    while todo:
        path = todo.popleft() if bfs else todo.pop()   # BFS: FIFO, DFS: LIFO
        node = path[-1]
        if node in visited:
            continue
        visited.add(node)
        explored += 1
        if node == end:
            return path, explored
        todo.extend(path + [n] for n in get_neighbours(maze, node) if n not in visited)
    return None, explored


def maze_bfs(maze, start, end):
    return search(maze, start, end, True)


def maze_dfs(maze, start, end):
    return search(maze, start, end, False)
```

```python
def timed(func, *args):
    t = time.perf_counter()
    path, nodes = func(*args)
    return path, nodes, (time.perf_counter() - t) * 1000


def compare_performance(results):
    print("\n--- PERFORMANCE COMPARISON ---")
    print("Algorithm | Steps | Total Cost | Nodes Explored | Time (ms)")
    print("-" * 59)
    steps = {}
    for name, (path, nodes, ms) in results.items():
        steps[name] = len(path) - 1 if path else 0      # every step costs 1
        print(f"{name:<9} | {steps[name]:<5} | {steps[name]:<10} | {nodes:<14} | {ms:.4f}")

    if results["BFS"][0] and results["DFS"][0]:
        if steps["BFS"] < steps["DFS"]:
            print("Verdict: BFS found a cheaper path.")
        elif steps["DFS"] < steps["BFS"]:
            print("Verdict: DFS found a cheaper path.")
        else:
            print("Verdict: Both paths have the same cost.")
    else:
        print("Verdict: No path exists between the given points.")
```

```python
while True:
    print("\n===== BFS vs DFS ASSIGNMENT =====")
    print("1. Maze Navigation")
    print("2. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        maze, start, end = get_maze()
        if start is None or end is None:
            print("Maze must contain both S and E!")
            continue
        print("Start:", start, "End:", end)

        results = {"BFS": timed(maze_bfs, maze, start, end),
                   "DFS": timed(maze_dfs, maze, start, end)}
        for name, (path, nodes, ms) in results.items():
            print(f"\n{name} Maze Path:", path)
            show_maze_path(maze, path)
        compare_performance(results)

    elif choice == "2":
        print("Exiting...")
        break
    else:
        print("Invalid choice, try again.")
```
