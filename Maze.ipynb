```python
import time
from collections import deque
```

```python
def get_maze():
    print("\n--- MAZE INPUT ---")
    print("Use 0 = open path, 1 = wall, S = start, E = end")

    rows = int(input("Enter number of rows: "))
    cols = int(input("Enter number of columns: "))

    maze = []
    for i in range(rows):
        row = input(f"Row {i+1}: ").replace(" ", "")

        # keep asking until the row has the right length
        while len(row) != cols:
            print(f"Row must have exactly {cols} characters, try again.")
            row = input(f"Row {i+1}: ").replace(" ", "")

        maze.append(row)

    # Find Start (S) and End (E)
    start = None
    end = None

    for r in range(rows):
        for c in range(cols):
            if maze[r][c] == "S":
                start = (r, c)
            elif maze[r][c] == "E":
                end = (r, c)

    return maze, start, end


def get_neighbours(maze, pos):
    r, c = pos
    result = []

    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr = r + dr
        nc = c + dc

        if 0 <= nr < len(maze) and 0 <= nc < len(maze[nr]):
            if maze[nr][nc] != "1":      # 1 is a wall
                result.append((nr, nc))

    return result


def show_maze_path(maze, path):
    if not path:
        print("No path found!")
        return

    grid = [list(row) for row in maze]

    for r, c in path:
        if grid[r][c] == "0":
            grid[r][c] = "*"

    for row in grid:
        print("".join(row))
```

```python
def maze_bfs(maze, start, end):
    queue = deque()
    queue.append([start])      # queue stores paths
    visited = set()
    visited.add(start)
    explored = 0

    while queue:
        path = queue.popleft()     # take from FRONT (FIFO)
        node = path[-1]
        explored += 1

        if node == end:
            return path, explored

        for n in get_neighbours(maze, node):
            if n not in visited:
                visited.add(n)
                queue.append(path + [n])

    return None, explored


def maze_dfs(maze, start, end):
    stack = deque()
    stack.append([start])      # stack stores paths
    visited = set()
    explored = 0

    while stack:
        path = stack.pop()         # take from BACK (LIFO)
        node = path[-1]

        if node in visited:
            continue
        visited.add(node)
        explored += 1

        if node == end:
            return path, explored

        for n in get_neighbours(maze, node):
            if n not in visited:
                stack.append(path + [n])

    return None, explored
```

```python
def compare_performance(bfs_path, bfs_nodes, bfs_time, bfs_cost,
                        dfs_path, dfs_nodes, dfs_time, dfs_cost):

    bfs_steps = len(bfs_path) - 1 if bfs_path else 0
    dfs_steps = len(dfs_path) - 1 if dfs_path else 0

    print("\n--- PERFORMANCE COMPARISON ---")
    print("Algorithm | Steps | Total Cost | Nodes Explored | Time (ms)")
    print("-----------------------------------------------------------")
    print(f"BFS       | {bfs_steps:<5} | {bfs_cost:<10} | {bfs_nodes:<14} | {bfs_time:.4f}")
    print(f"DFS       | {dfs_steps:<5} | {dfs_cost:<10} | {dfs_nodes:<14} | {dfs_time:.4f}")

    if bfs_path and dfs_path:
        if bfs_cost < dfs_cost:
            print("Verdict: BFS found a cheaper path.")
        elif dfs_cost < bfs_cost:
            print("Verdict: DFS found a cheaper path.")
        else:
            print("Verdict: Both paths have the same cost.")
    else:
        print("Verdict: No path exists between the given points.")
```

```python
while True:
    print("\n===== BFS vs DFS ASSIGNMENT =====")
    print("1. Maze Navigation")
    print("2. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        maze, start, end = get_maze()

        if start is None or end is None:
            print("Maze must contain both S and E!")
            continue

        print("Start:", start, "End:", end)

        t1 = time.perf_counter()
        bfs_path, bfs_nodes = maze_bfs(maze, start, end)
        bfs_time = (time.perf_counter() - t1) * 1000

        t2 = time.perf_counter()
        dfs_path, dfs_nodes = maze_dfs(maze, start, end)
        dfs_time = (time.perf_counter() - t2) * 1000

        print("\nBFS Maze Path:", bfs_path)
        show_maze_path(maze, bfs_path)

        print("\nDFS Maze Path:", dfs_path)
        show_maze_path(maze, dfs_path)

        # in a maze every step costs 1
        bfs_cost = len(bfs_path) - 1 if bfs_path else 0
        dfs_cost = len(dfs_path) - 1 if dfs_path else 0

        compare_performance(bfs_path, bfs_nodes, bfs_time, bfs_cost,
                            dfs_path, dfs_nodes, dfs_time, dfs_cost)

    elif choice == "2":
        print("Exiting...")
        break

    else:
        print("Invalid choice, try again.")
```
